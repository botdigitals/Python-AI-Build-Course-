```python
import random

# A tiny text. Our predictor will learn from it.
text = "the cat sat on the mat the cat ate the fish the dog sat on the rug"
words = text.split()

# Step 1: count which word comes after which word
follows = {}
for current, nxt in zip(words, words[1:]):
    if current not in follows:
        follows[current] = []
    follows[current].append(nxt)

print("Words that came after 'the':", follows["the"])

# Step 2: write new text by picking a next word again and again
random.seed(5)
word = "the"
sentence = [word]
for _ in range(8):
    if word not in follows:
        break
    word = random.choice(follows[word])
    sentence.append(word)

print("New text:", " ".join(sentence))
```
